# Phase 2 — Exploratory Data Analysis (EDA)
**Customer Churn Prediction Pipeline**

In this notebook, we perform exploratory data analysis on the  dataset to analyze predictive relationships, inspect distributions, and identify patterns before training our machine learning model.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../data/Telco-Customer-Churn.csv')

# Prepare TotalCharges for EDA (convert to float, coercing errors to NaN)
df['TotalCharges_num'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print(f'Dataset Shape: {df.shape}')
```

## 1. Target Variable Distribution

```
plt.figure(figsize=(6, 4))
ax = sns.countplot(x='Churn', data=df, hue='Churn', palette=['#2b5c8f', '#d95f02'], legend=False)
plt.title('Target Distribution: Customer Churn', fontsize=12, fontweight='bold')
plt.xlabel('Churn Status')
plt.ylabel('Customer Count')
total = len(df)
for p in ax.patches:
    height = p.get_height()
    pct = height / total * 100
    ax.annotate(f'{height}
({pct:.1f}%)', (p.get_x() + p.get_width() / 2., height / 2),
                ha='center', va='center', fontsize=10, color='white', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 2. Contract Type vs Churn Rate

```
plt.figure(figsize=(7, 4.5))
contract_churn = df.groupby('Contract')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100).reset_index()
ax = sns.barplot(x='Contract', y='Churn', data=contract_churn, hue='Contract', palette='Blues_r', legend=False)
plt.title('Observed Churn Rate by Contract Type (%)', fontsize=12, fontweight='bold')
plt.xlabel('Contract Type')
plt.ylabel('Churn Rate (%)')
for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:.1f}%', (p.get_x() + p.get_width() / 2., height + 1),
                ha='center', va='bottom', fontsize=10, fontweight='bold')
plt.ylim(0, 50)
plt.tight_layout()
plt.show()
```

## 3. Tenure vs Churn

```
plt.figure(figsize=(7, 4))
sns.boxplot(x='Churn', y='tenure', data=df, hue='Churn', palette=['#2b5c8f', '#d95f02'], legend=False)
plt.title('Tenure Distribution by Churn Status', fontsize=12, fontweight='bold')
plt.xlabel('Churn Status')
plt.ylabel('Tenure (Months)')
plt.tight_layout()
plt.show()

print('Tenure Summary Statistics by Churn:')
print(df.groupby('Churn')['tenure'].describe())
```

## 4. Monthly Charges vs Churn

```
plt.figure(figsize=(7, 4))
sns.boxplot(x='Churn', y='MonthlyCharges', data=df, hue='Churn', palette=['#2b5c8f', '#d95f02'], legend=False)
plt.title('Monthly Charges Distribution by Churn Status', fontsize=12, fontweight='bold')
plt.xlabel('Churn Status')
plt.ylabel('Monthly Charges ($)')
plt.tight_layout()
plt.show()

print('Monthly Charges Summary Statistics by Churn:')
print(df.groupby('Churn')['MonthlyCharges'].describe())
```

## 5. Internet Service vs Churn Rate

```
plt.figure(figsize=(7, 4.5))
internet_churn = df.groupby('InternetService')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100).reset_index()
ax = sns.barplot(x='InternetService', y='Churn', data=internet_churn, hue='InternetService', palette='Oranges_r', legend=False)
plt.title('Observed Churn Rate by Internet Service (%)', fontsize=12, fontweight='bold')
plt.xlabel('Internet Service Type')
plt.ylabel('Churn Rate (%)')
for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:.1f}%', (p.get_x() + p.get_width() / 2., height + 1),
                ha='center', va='bottom', fontsize=10, fontweight='bold')
plt.ylim(0, 50)
plt.tight_layout()
plt.show()
```

## 6. Payment Method vs Churn Rate

```
plt.figure(figsize=(8.5, 4.5))
pay_churn = df.groupby('PaymentMethod')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100).reset_index().sort_values('Churn', ascending=False)
ax = sns.barplot(x='Churn', y='PaymentMethod', data=pay_churn, hue='PaymentMethod', palette='Purples_r', legend=False)
plt.title('Observed Churn Rate by Payment Method (%)', fontsize=12, fontweight='bold')
plt.xlabel('Churn Rate (%)')
plt.ylabel('Payment Method')
for p in ax.patches:
    width = p.get_width()
    ax.annotate(f'{width:.1f}%', (width + 0.8, p.get_y() + p.get_height()/2.),
                ha='left', va='center', fontsize=10, fontweight='bold')
plt.xlim(0, 55)
plt.tight_layout()
plt.show()
```

## 7. Numerical Feature Correlation

```
plt.figure(figsize=(6, 5))
df_num = df[['tenure', 'MonthlyCharges', 'TotalCharges_num']].copy()
df_num['Churn'] = (df['Churn'] == 'Yes').astype(int)
sns.heatmap(df_num.corr(), annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1, cbar=True)
plt.title('Correlation Heatmap', fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
```
